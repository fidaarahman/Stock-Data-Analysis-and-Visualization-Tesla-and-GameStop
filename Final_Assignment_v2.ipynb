## Define Graphing Function


In this section, we define the function `make_graph`. **You don't have to know how the function works, you should only care about the inputs. It takes a dataframe with stock data (dataframe must contain Date and Close columns), a dataframe with revenue data (dataframe must contain Date and Revenue columns), and the name of the stock.**


````python
def make_graph(stock_data, revenue_data, stock):
    fig = make_subplots(rows=2, cols=1, shared_xaxes=True, subplot_titles=("Historical Share Price", "Historical Revenue"), vertical_spacing = .3)
    stock_data_specific = stock_data[stock_data.Date <= '2021-06-14']
    revenue_data_specific = revenue_data[revenue_data.Date <= '2021-04-30']
    fig.add_trace(go.Scatter(x=pd.to_datetime(stock_data_specific.Date, infer_datetime_format=True), y=stock_data_specific.Close.astype("float"), name="Share Price"), row=1, col=1)
    fig.add_trace(go.Scatter(x=pd.to_datetime(revenue_data_specific.Date, infer_datetime_format=True), y=revenue_data_specific.Revenue.astype("float"), name="Revenue"), row=2, col=1)
    fig.update_xaxes(title_text="Date", row=1, col=1)
    fig.update_xaxes(title_text="Date", row=2, col=1)
    fig.update_yaxes(title_text="Price ($US)", row=1, col=1)
    fig.update_yaxes(title_text="Revenue ($US Millions)", row=2, col=1)
    fig.update_layout(showlegend=False,
    height=900,
    title=stock,
    xaxis_rangeslider_visible=True)
    fig.show()
    from IPython.display import display, HTML
    fig_html = fig.to_html()
    display(HTML(fig_html))
````

## Question 2: Use Webscraping to Extract Tesla Revenue Data


Use the `requests` library to download the webpage https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/revenue.htm Save the text of the response as a variable named `html_data`.


````python
import pandas as pd
from IPython.display import display

REVENUE_URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/revenue.htm"

def get_quarterly_revenue(url: str, match_text: str) -> pd.DataFrame:
    """
    Returns a clean DataFrame with columns ['Date', 'Revenue'] from the course revenue page.
    Handles varying headers and avoids invalid escape sequences.
    """
    # Read the specific table by matching its heading text
    tables = pd.read_html(url, match=match_text, flavor="bs4")
    if not tables:
        raise RuntimeError(f"No table matched: {match_text}")

    raw = tables[0].copy()

    # Auto-pick the first two columns (header names can vary)
    df = raw.iloc[:, :2].copy()

    # Standardize column names to avoid KeyError
    df.columns = ["Date", "Revenue"]

    # Drop any header-like rows that got parsed as data (e.g., if 'Revenue' text leaked into rows)
    # Keep only rows where Revenue contains at least one digit
    df = df[df["Revenue"].astype(str).str.contains(r"\d", na=False)].copy()

    # Clean revenue: remove $ and commas (no invalid escapes)
    df["Revenue"] = (
        df["Revenue"]
        .astype(str)
        .str.replace(r"[$,]", "", regex=True)  # <- avoids SyntaxWarning
        .str.strip()
    )

    # Convert types
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
    df["Revenue"] = pd.to_numeric(df["Revenue"], errors="coerce")

    # Final clean
    df = df.dropna(subset=["Date", "Revenue"]).reset_index(drop=True)
    return df

# --- Q2: Tesla ---
tesla_revenue = get_quarterly_revenue(REVENUE_URL, "Tesla Quarterly Revenue")
display(tesla_revenue.head())

# --- Q4: GameStop ---
gme_revenue = get_quarterly_revenue(REVENUE_URL, "GameStop Quarterly Revenue")
display(gme_revenue.head())
````

````python
# Install necessary libraries if not already installed
!pip install yfinance plotly

import yfinance as yf
import pandas as pd
import plotly.graph_objects as go

# Step 1: Extract GameStop Stock Data using yfinance
gme = yf.Ticker('GME')
gme_data = gme.history(period="5y")

# Step 2: Extract GameStop Annual Revenue Data (you can update this if you have different values)
annual_revenue = {
    'Year': [2025, 2024, 2023, 2022, 2021, 2020, 2019, 2018, 2017, 2016, 2015, 2014, 2013, 2012, 2011],
    'Revenue (Millions USD)': [3823, 5273, 5927, 6011, 5090, 6466, 8285, 8547, 7965, 9364, 9296, 9040, 8887, 9551, 9474]
}

# Convert to DataFrame
gme_revenue = pd.DataFrame(annual_revenue)
gme_revenue['Year'] = pd.to_datetime(gme_revenue['Year'], format='%Y')

# Step 3: Resample GameStop Stock Data and Merge with Revenue Data
# Resample the stock data to yearly data (get last stock price for each year)
gme_yearly = gme_data.resample('Y').last()

# Create a 'Year' column for merging
gme_yearly['Year'] = gme_yearly.index.year

# Convert 'Year' in gme_revenue to int
gme_revenue['Year'] = gme_revenue['Year'].dt.year

# Merge the stock price data with the annual revenue data
merged_data = pd.merge(gme_yearly[['Close', 'Year']], gme_revenue[['Year', 'Revenue (Millions USD)']], on='Year', how='inner')

# Step 4: Create the Dashboard using Plotly
fig = go.Figure()

# Plot the GameStop stock price data as a line graph
fig.add_trace(go.Scatter(x=merged_data['Year'], y=merged_data['Close'], mode='lines', name='Stock Price (USD)', line=dict(color='blue')))

# Plot the GameStop annual revenue as a bar chart
fig.add_trace(go.Bar(x=merged_data['Year'], y=merged_data['Revenue (Millions USD)'], name='Revenue (Millions USD)', marker_color='orange'))

# Customize the layout
fig.update_layout(
    title="GameStop Stock and Annual Revenue Dashboard",
    xaxis_title="Year",
    yaxis_title="Stock Price (USD)",
    barmode='group',  # Group the bars for side-by-side comparison
    template='plotly_dark',
    legend_title="Metrics",
    xaxis=dict(
        tickmode='array',
        tickvals=merged_data['Year'],
        ticktext=[str(year) for year in merged_data['Year']]
    ),
)

# Show the interactive plot
fig.show()
````
